```python
# ==========================================
# TASK 1: STORE CUSTOMER ORDERS
# ==========================================

# 1.a Create a list of customer names
customer_names = ["Alice", "Bob", "Charlie", "David", "Emma", "Frank"]

# 1.b Store each customer's order details (customer name, product, price, category) as tuples inside a list
orders = [
    ("Alice", "Laptop", 800.00, "Electronics"),
    ("Alice", "Mouse", 25.00, "Electronics"),
    ("Bob", "T-Shirt", 20.00, "Clothing"),
    ("Bob", "Jeans", 45.00, "Clothing"),
    ("Bob", "Laptop", 800.00, "Electronics"),
    ("Charlie", "Coffee Maker", 60.00, "Home Essentials"),
    ("Charlie", "Blender", 35.00, "Home Essentials"),
    ("David", "Headphones", 120.00, "Electronics"),
    ("David", "Socks", 10.00, "Clothing"),
    ("Emma", "Water Bottle", 15.00, "Home Essentials"),
    ("Frank", "Smartphone", 500.00, "Electronics"),
    ("Frank", "Jacket", 85.00, "Clothing")
]
```

```python
#Task 1.c Use a dictionary where keys are customer names and values are lists of ordered products
customer_order_dict = {}
for name, product, price, category in orders:
    if name not in customer_order_dict:
        customer_order_dict[name] =[product]
    else:
        customer_order_dict[name].append(product)
print(customer_order_dict)
```

```python
# 2. Classify products by category
#2.a Use a dictionary to map each product to its respective category
product_category_dict = {}
for name, product, price, category in orders:
    product_category_dict[product] = category
print(product_category_dict)
```

```python
#2.b. Create a set of unique product categories
product_category_set = set(product_category_dict.values())
#2.c • Display all available product categories
print(product_category_set)
```

```python
#3. Analyze customer orders
#3.a. Use a loop to calculate the total amount each customer spends
customer_total_spends = {}

for name, product, price, category in orders:
    if name not in customer_total_spends:
        customer_total_spends[name] = price
    else:
        customer_total_spends[name] = customer_total_spends[name] + price

print(customer_total_spends)
```

```python
#3.b If the total purchase value is above $100, classify the customer as a high-value buyer
#3.c If it is between $50 and $100, classify the customer as a moderate buyer
#3.d If it is below $50, classify them as a low-value buyer

customer_classification = {}
for name, total_spends in customer_total_spends.items():
    if(total_spends > 100) :
        customer_classification[name] = 'High Value Buyer'
    elif (total_spends >50 and total_spends <=100):
        customer_classification[name] = 'Moderate Value Buyer'
    else:
        customer_classification[name] = 'Low Value Buyer'

print(customer_classification)
```

```python
#4 Generate business insights
#4.a. Calculate the total revenue per product category and store it in a dictionary
total_revenue_per_category = {}
for name, product, price, category in orders:
    if category not in total_revenue_per_category:
        total_revenue_per_category[category] = price
    else:
        total_revenue_per_category[category] = total_revenue_per_category[category]+price

print(total_revenue_per_category)
```

```python
#4.b.Extract unique products from all orders using a set
unique_products_set = {}
unique_products_set = set([item for list in customer_order_dict.values() for item in list])

print(unique_products_set)
```

```python
#4.c Use a list comprehension to find all customers who purchased electronics
customer_who_purchased_electronics = [name for name, product, price, category in orders if (category == 'Electronics')]
print(set(customer_who_purchased_electronics))
```

```python
#4.d Identify the top three highest-spending customers using sorting
sorted_data = sorted(customer_total_spends.items(), key=lambda item: item[1], reverse=True)
top_three_spending_csutomer = sorted_data[:3:]

print(top_three_spending_csutomer)
```

```python
#5. Organize and display data
#5.a Print a summary of each customer’s total spending and their classification
for name in customer_names:
    print(f'customer name : {name:<10} | total spending : {customer_total_spends[name]:<7.2f} | customer classification : {customer_classification[name]} ')
```

```python
#5.b Use set operations to find customers who purchased from multiple categories
electronics_buyers = {name for name, _, _, category in orders if category == "Electronics"}
clothing_buyers = {name for name, _, _, category in orders if category == "Clothing"}
home_buyers = {name for name, _, _, category in orders if category == "Home Essentials"}

elec_and_cloth = electronics_buyers & clothing_buyers
elec_and_home  = electronics_buyers & home_buyers
cloth_and_home = clothing_buyers & home_buyers

# Union all the overlapping sets together to get all multi-category buyers
multi_category_buyers = elec_and_cloth.union(elec_and_home).union(cloth_and_home)

print(f"Multi-Category Buyers: {multi_category_buyers}")
```

```python
#5.c Identify common customers who bought both electronics and clothing
elec_clothing_shoppers = electronics_buyers.intersection(clothing_buyers)
print(elec_clothing_shoppers)
```

```python
# Insight generation
#• Extract the high-value customers

high_value_cust = [name for name, classification in customer_classification.items() if classification == 'High Value Buyer']
print(high_value_cust)
```
